```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from vibrastruct import *
```

## 1. Single-mass resonance
Predict the natural frequency and compare the effect of damping.

```python
m,k=250.,1e6
f=np.linspace(0,30,2401)
for z in [0.01,0.03,0.10]:
    plt.plot(f,1000*100*np.abs(receptance(m,k,z,2*np.pi*f)),label=str(z))
plt.xlabel("Frequency [Hz]"); plt.ylabel("Amplitude [mm]"); plt.legend(); plt.show()
```

## 2. Coupled modes
The first spring connects to ground, and the final mass has a free end.

```python
M,K=chain([250,180,120],[1e6,8e5,6e5])
w,P=modes(M,K)
print("Frequencies [Hz]:",w/(2*np.pi))
print("Mass orthogonality:\n",P.T@M@P)
```

## 3. Free vibration and energy
Displace mass 3 by 2 mm, release from rest, and compare energy with the undamped case.

```python
t=np.linspace(0,3,3001)
C=rayleigh(M,K,0.03)
x,v=response(M,K,C,t,[0,0,0.002],[0,0,0])
plt.plot(t,x*1000); plt.xlabel("Time [s]"); plt.ylabel("Displacement [mm]"); plt.show()
plt.plot(t,energy(M,K,x,v)); plt.xlabel("Time [s]"); plt.ylabel("Energy [J]"); plt.show()
```
